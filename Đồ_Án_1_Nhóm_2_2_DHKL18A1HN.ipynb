TÓM TẮT NỘI DUNG ĐỒ ÁN

**1. Giới thiệu**

Kết quả học tập của học sinh chịu tác động của nhiều yếu tố, gồm năng lực cá nhân, môi trường giáo dục, điều kiện gia đình và đặc điểm nhân khẩu học. Phân tích dữ liệu về các yếu tố này giúp nhà trường, giáo viên và phụ huynh hiểu rõ hơn sự khác biệt giữa các nhóm học sinh, từ đó lựa chọn biện pháp hỗ trợ phù hợp. Trong số đó, nhóm sắc tộc/chủng tộc và trình độ học vấn của phụ huynh là hai đặc điểm thường được quan tâm khi nghiên cứu về thành tích học tập.

Đồ án sử dụng bộ dữ liệu Students Performance in Exams trên Kaggle, gồm khoảng 1.000 học sinh với 8 cột dữ liệu, chia thành hai nhóm:

Thông tin học sinh và gia đình (5 biến phân loại): gender (giới tính); race/ethnicity (nhóm sắc tộc/chủng tộc, được ẩn danh thành Group A đến Group E); parental level of education (trình độ học vấn của phụ huynh); lunch (loại bữa trưa, thường được dùng như chỉ báo gián tiếp về điều kiện kinh tế); và test preparation course (tình trạng tham gia khóa chuẩn bị thi).

Kết quả học tập (3 biến số): math score, reading score và writing score, lần lượt là điểm Toán, Đọc hiểu và Viết, theo thang điểm 0-100.

Bộ dữ liệu này phổ biến trong các bài tập phân tích, nhưng nhiều nguồn cho rằng dữ liệu mang tính giả lập và các nhóm sắc tộc đã được ẩn danh. Vì vậy, kết quả của đồ án chủ yếu có giá trị minh họa phương pháp phân tích, chưa thể khái quát hóa cho toàn bộ học sinh trong thực tế.

Mục tiêu của đồ án là thăm dò, phân tích và trực quan hóa dữ liệu (Thông qua Exploratory Data Analysis hay EDA) nhằm làm rõ mối liên hệ giữa race/ethnicity, parental level of education và kết quả học tập của học sinh. Cụ thể, nhóm tìm câu trả lời cho ba câu hỏi: (1) điểm thi có khác biệt đáng kể giữa các nhóm sắc tộc từ Group A đến Group E hay không; (2) điểm thi có khác biệt giữa các nhóm học sinh có phụ huynh với trình độ học vấn khác nhau hay không; và (3) những khác biệt này còn giữ nguyên hay thay đổi khi xét thêm các biến lunch, test preparation course và gender.

Để trả lời các câu hỏi trên, nhóm kiểm tra chất lượng dữ liệu (dữ liệu thiếu, trùng lặp, sai định dạng) và xem xét các giá trị ngoại lệ trong điểm thi. Nhóm xây dựng biến điểm trung bình tổng hợp (average score) bằng trung bình cộng của math score, reading score và writing score để mô tả kết quả học tập chung. Biến này được sử dụng cho thống kê mô tả, trực quan hóa và kiểm định sự khác biệt giữa các nhóm; không sử dụng đồng thời ba điểm thành phần làm biến đầu vào để dự đoán nhóm được tạo trực tiếp từ average score, nhằm tránh rò rỉ dữ liệu (data leakage). Kết quả được trình bày qua các bảng thống kê mô tả, biểu đồ và kiểm định thống kê phù hợp.

Cần lưu ý rằng đồ án chỉ mô tả và so sánh dữ liệu thống kê. Các khác biệt về điểm số, nếu có, không đủ để kết luận sắc tộc hay học vấn của phụ huynh là nguyên nhân trực tiếp, vì chúng có thể phản ánh các yếu tố khác như điều kiện kinh tế hay điều kiện học tập. Việc so sánh giữa các nhóm sắc tộc nhằm mục đích mô tả dữ liệu, không nhằm đánh giá hay quy kết năng lực của bất kỳ nhóm nào.
Phần còn lại của báo cáo gồm: cơ sở lý thuyết ngắn gọn, quá trình tiền xử lý dữ liệu, kết quả phân tích khám phá, thảo luận và kết luận.

4.1. Thống kê suy luận

-Mục đích là kiểm tra liệu sự khác biệt về average score giữa các nhóm có ý nghĩa thống kê hay chỉ xuất hiện do biến động ngẫu nhiên của dữ liệu.

-Cách thực hiện:
+ Kiểm tra các điều kiện trước khi thực hiện ANOVA một yếu tố để so sánh average score giữa các nhóm sắc tộc: các quan sát độc lập, phần dư gần phân phối chuẩn trong từng nhóm và phương sai giữa các nhóm tương đối đồng nhất.
+ Thực hiện tương tự khi so sánh average score giữa các nhóm trình độ học vấn của phụ huynh. Có thể dùng biểu đồ Q-Q hoặc kiểm định phù hợp để đánh giá tính chuẩn; dùng Levene’s test để kiểm tra tính đồng nhất phương sai.
+ Đặt mức ý nghĩa thống kê α = 0,05. Nếu điều kiện về phương sai không được đáp ứng, cân nhắc Welch ANOVA; nếu dữ liệu không đáp ứng phù hợp giả định phân phối của ANOVA, cân nhắc kiểm định phi tham số Kruskal–Wallis.
+ Nếu kiểm định tổng thể có ý nghĩa thống kê, thực hiện kiểm định hậu nghiệm phù hợp: Tukey HSD sau ANOVA chuẩn; Games–Howell khi phương sai không đồng nhất; hoặc Dunn sau Kruskal–Wallis, kèm điều chỉnh khi so sánh nhiều cặp.

-Mục đích:
+ Xác định liệu nhóm sắc tộc và trình độ học vấn của phụ huynh có liên quan đến sự khác biệt trong điểm trung bình của học sinh hay không.
+ Cung cấp cơ sở thống kê cho các nhận xét và kết luận từ phần trực quan hóa dữ liệu.

4.3. Phân tích nguyên nhân

-Mục đích là tìm hiểu các yếu tố có thể liên quan đến sự khác biệt trong kết quả học tập của học sinh.

-Cách thực hiện:
+ Phân tích sự khác biệt về điểm trung bình giữa các nhóm sắc tộc.
+ Phân tích sự khác biệt về điểm trung bình giữa các nhóm trình độ học vấn của phụ huynh.
+ Kết hợp hai yếu tố sắc tộc và trình độ học vấn của phụ huynh để quan sát sự khác biệt giữa các nhóm.
+ Phân tích tương quan giữa điểm Toán, Đọc và Viết để xem xét mối liên hệ giữa các điểm thành phần. Vì average score được tính trực tiếp từ ba điểm này, không diễn giải tương quan giữa average score và các điểm thành phần như một phát hiện độc lập hay bằng chứng dự báo độc lập.
+ Sử dụng kết quả ANOVA, Post-hoc và tương quan để xác định những yếu tố có mối liên hệ đáng chú ý với kết quả học tập.

-Mục đích:
+ Làm rõ những yếu tố có mối liên hệ với sự khác biệt về kết quả học tập.
+ Hỗ trợ giải thích tại sao kết quả học tập có sự khác biệt giữa các nhóm học sinh.
+ Làm cơ sở cho bước tiếp theo là xây dựng Decision Tree hoặc Logistic Regression để phân loại học sinh theo nhóm kết quả học tập, với điều kiện các biến đầu vào không chứa trực tiếp hoặc gián tiếp thông tin dùng để tạo nhãn mục tiêu.


5.Chủ đề nâng cao: Phân tích mô hình dự đoán kết quả học tập
-Mục đích: Xây dựng mô hình phân loại nhóm kết quả học tập được xác định từ average score, đồng thời tránh rò rỉ dữ liệu (data leakage).

-Nhãn mục tiêu: tạo từ average score theo ngưỡng đã xác định trước: dưới 60 điểm là nhóm cần hỗ trợ; từ 60 đến dưới 80 điểm là nhóm trung bình; từ 80 điểm trở lên là nhóm kết quả cao. Các ngưỡng này là quy ước phân tích của đồ án.

-Phương pháp và dữ liệu đầu vào: Áp dụng Decision Tree hoặc Logistic Regression, sử dụng các biến thông tin có sẵn như race/ethnicity, parental level of education, gender, lunch và test preparation course. Không sử dụng math score, reading score, writing score hoặc average score làm biến đầu vào vì chúng trực tiếp tạo nên nhãn mục tiêu; việc dùng chúng sẽ gây data leakage và làm kết quả đánh giá không đáng tin cậy.

-Đánh giá mô hình: Chia dữ liệu thành tập huấn luyện và kiểm tra trước khi huấn luyện; dùng Accuracy, Precision, Recall, F1-score và Confusion Matrix. Nếu có thể, sử dụng phân tầng theo nhãn khi chia tập dữ liệu và chỉ thực hiện các bước học từ dữ liệu trên tập huấn luyện.

-Phân tích kết quả: So sánh kết quả dự đoán trên tập kiểm tra với nhãn thực tế; nêu rõ giới hạn của mô hình và không diễn giải các biến đầu vào là nguyên nhân trực tiếp của kết quả học tập.

Ý nghĩa: Bổ sung góc nhìn dự đoán bằng học máy. Độ chính xác chỉ được xem là có ý nghĩa khi đánh giá trên dữ liệu chưa dùng để huấn luyện và không có rò rỉ dữ liệu.

Kết quả học tập của học sinh chịu tác động của nhiều yếu tố, gồm năng lực cá nhân, môi trường giáo dục, điều kiện gia đình và đặc điểm nhân khẩu học. Phân tích dữ liệu về các yếu tố này giúp nhà trường, giáo viên và phụ huynh hiểu rõ hơn sự khác biệt giữa các nhóm học sinh, từ đó lựa chọn biện pháp hỗ trợ phù hợp. Trong số đó, nhóm sắc tộc/chủng tộc và trình độ học vấn của phụ huynh là hai đặc điểm thường được quan tâm khi nghiên cứu về thành tích học tập.

Đồ án sử dụng bộ dữ liệu Students Performance in Exams trên Kaggle, gồm khoảng 1.000 học sinh với 8 cột dữ liệu, chia thành hai nhóm: Thông tin học sinh và gia đình (5 biến phân loại): gender (giới tính); race/ethnicity (nhóm sắc tộc/chủng tộc, được ẩn danh thành Group A đến Group E); parental level of education (trình độ học vấn của phụ huynh); lunch (loại bữa trưa, thường được dùng như chỉ báo gián tiếp về điều kiện kinh tế); và test preparation course (tình trạng tham gia khóa chuẩn bị thi). Kết quả học tập (3 biến số): math score, reading score và writing score, lần lượt là điểm Toán, Đọc hiểu và Viết, theo thang điểm 0-100. Bộ dữ liệu này phổ biến trong các bài tập phân tích, nhưng nhiều nguồn cho rằng dữ liệu mang tính giả lập và các nhóm sắc tộc đã được ẩn danh. Vì vậy, kết quả của đồ án chủ yếu có giá trị minh họa phương pháp phân tích, chưa thể khái quát hóa cho toàn bộ học sinh trong thực tế.

Mục tiêu của đồ án là thăm dò, phân tích và trực quan hóa dữ liệu (Thông qua Exploratory Data Analysis hay EDA) nhằm làm rõ mối liên hệ giữa race/ethnicity, parental level of education và kết quả học tập của học sinh. Cụ thể, nhóm tìm câu trả lời cho ba câu hỏi: (1) điểm thi có khác biệt đáng kể giữa các nhóm sắc tộc từ Group A đến Group E hay không; (2) điểm thi có khác biệt giữa các nhóm học sinh có phụ huynh với trình độ học vấn khác nhau hay không; và (3) những khác biệt này còn giữ nguyên hay thay đổi khi xét thêm các biến lunch, test preparation course và gender. Để trả lời các câu hỏi trên, nhóm kiểm tra chất lượng dữ liệu (dữ liệu thiếu, trùng lặp, sai định dạng), mã hóa các biến phân loại và xem xét các giá trị ngoại lệ trong điểm thi. Nhóm cũng xây dựng thêm biến điểm trung bình tổng hợp (average score), tính bằng trung bình cộng của math score, reading score và writing score, để đánh giá kết quả học tập chung. Kết quả được trình bày qua các bảng thống kê mô tả và biểu đồ, kèm kiểm định thống kê để xác định khác biệt có ý nghĩa hay không. Cần lưu ý rằng đồ án chỉ mô tả và so sánh dữ liệu thống kê. Các khác biệt về điểm số, nếu có, không đủ để kết luận sắc tộc hay học vấn của phụ huynh là nguyên nhân trực tiếp, vì chúng có thể phản ánh các yếu tố khác như điều kiện kinh tế hay điều kiện học tập. Việc so sánh giữa các nhóm sắc tộc nhằm mục đích mô tả dữ liệu, không nhằm đánh giá hay quy kết năng lực của bất kỳ nhóm nào. Phần còn lại của báo cáo gồm: cơ sở lý thuyết ngắn gọn, quá trình tiền xử lý dữ liệu, kết quả phân tích khám phá, thảo luận và kết luận
